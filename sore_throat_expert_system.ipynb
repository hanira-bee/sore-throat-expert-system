```
# Anggota Kelompok :
# Duta Alwi Prasetya (17230774)
# Putri Ramadhani (17230496)
# Chiara Nurtabina (17230159)
"""
Sistem Pakar Sederhana untuk Diagnosis Awal Radang Tenggorokan
-------------------------------------------------------------
Program ini menggunakan metode forward chaining dengan sistem skor
untuk memberikan diagnosis awal berdasarkan gejala yang dimasukkan pengguna.

PENTING:
Program ini HANYA alat bantu edukasi dan TIDAK MENGGANTIKAN
konsultasi medis profesional. Diagnosis pasti harus dilakukan oleh dokter.
"""

def get_user_input(question):
    """Fungsi untuk mendapatkan jawaban ya/tidak yang valid dari pengguna."""
    while True:
        answer = input(question + " (y/n): ").lower().strip()
        if answer in ['y', 'yes', 'n', 'no']:
            return answer.startswith('y')
        else:
            print("Input tidak valid. Mohon jawab dengan 'y' (ya) atau 'n' (tidak).")

def expert_system_sore_throat():
    """Fungsi utama untuk menjalankan sistem pakar."""

    # Inisialisasi basis pengetahuan (gejala dan skornya)
    # Skor positif menandakan gejala yang mendukung diagnosis radang tenggorokan (bakteri)
    # Skor negatif menandakan gejala yang kurang mendukung (lebih ke arah virus)
    knowledge_base = {
        'nyeri_telan_parah': 3,
        'bintik_putih_amandel': 4, # Gejala yang sangat spesifik
        'demam_tinggi': 2,         # Demam di atas 38°C
        'pembengkakan_kelenjar': 2,
        'sakit_kepala': 1,
        'mual_muntah': 1,
        'batuk': -2,               # Gejala kuat yang mengarah ke infeksi virus
        'hidung_meler': -1
    }

    score = 0

    print("="*50)
    print("   SISTEM PAKAR DIAGNOSIS AWAL RADANG TENGGOROKAN   ")
    print("="*50)
    print("\nJawab pertanyaan berikut untuk menganalisis gejala Anda.")
    print("PERINGATAN: Hasil dari sistem ini tidak menggantikan diagnosis dokter.\n")

    # Proses akuisisi pengetahuan dari pengguna (tanya jawab)
    if get_user_input("Apakah Anda mengalami nyeri saat menelan yang cukup parah?"):
        score += knowledge_base['nyeri_telan_parah']

    if get_user_input("Apakah Anda melihat bintik-bintik putih pada amandel Anda?"):
        score += knowledge_base['bintik_putih_amandel']

    if get_user_input("Apakah suhu tubuh Anda di atas 38° Celsius?"):
        score += knowledge_base['demam_tinggi']

    if get_user_input("Apakah Anda merasakan adanya pembengkakan kelenjar getah bening di leher?"):
        score += knowledge_base['pembengkakan_kelenjar']

    if get_user_input("Apakah Anda juga mengalami sakit kepala?"):
        score += knowledge_base['sakit_kepala']

    if get_user_input("Apakah Anda merasa mual atau ingin muntah?"):
        score += knowledge_base['mual_muntah']

    # Gejala yang berlawanan dengan radang tenggorokan akibat bakteri
    print("\n--- Pertanyaan Berikut Penting Untuk Membedakan ---")
    if get_user_input("Apakah Anda mengalami batuk?"):
        score += knowledge_base['batuk']

    if get_user_input("Apakah Anda mengalami hidung meler atau tersumbat?"):
        score += knowledge_base['hidung_meler']


    # Mesin inferensi (inference engine) - Menganalisis skor akhir
    print("\n" + "="*50)
    print("                  HASIL ANALISA                  ")
    print("="*50)
    print(f"Total Skor Gejala Anda: {score}\n")

    if score >= 6:
        print("DIAGNOSIS: Kemungkinan TINGGI Anda terkena radang tenggorokan akibat infeksi bakteri (Strep Throat).")
        print("REKOMENDASI: Sangat disarankan untuk segera berkonsultasi dengan dokter untuk mendapatkan pemeriksaan dan resep antibiotik jika diperlukan.")
    elif 3 <= score <= 5:
        print("DIAGNOSIS: Kemungkinan SEDANG Anda terkena radang tenggorokan.")
        print("REKOMENDASI: Gejala Anda bisa disebabkan oleh bakteri atau virus. Pantau kondisi Anda. Jika tidak membaik dalam 2-3 hari atau gejala memburuk, segera temui dokter.")
    else:
        print("DIAGNOSIS: Kemungkinan KECIL Anda terkena radang tenggorokan akibat bakteri.")
        print("REKOMENDASI: Gejala Anda lebih mengarah pada infeksi virus biasa (seperti flu atau pilek). Istirahat yang cukup dan minum banyak cairan. Jika gejala berlanjut atau memburuk, tetap konsultasikan dengan dokter.")

    print("\nINGAT: Ini adalah analisa awal berdasarkan data terbatas. Konsultasi medis profesional adalah langkah yang paling tepat.")
    print("="*50)


if __name__ == '__main__':
    expert_system_sore_throat()
```
